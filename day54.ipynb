```python
%%writefile day54.c
#include <stdio.h>
#include <stdlib.h>

struct Node {
    int data;
    struct Node *left, *right;
};

struct Node* newNode(int d) {
    struct Node* n = malloc(sizeof(struct Node));
    n->data = d;
    n->left = n->right = NULL;
    return n;
}


struct Node* build(int a[], int n) {
    struct Node* t[n];
    for(int i=0;i<n;i++)
        t[i] = (a[i]==-1)? NULL : newNode(a[i]);

    for(int i=0;i<n;i++) {
        if(t[i]) {
            if(2*i+1<n) t[i]->left = t[2*i+1];
            if(2*i+2<n) t[i]->right = t[2*i+2];
        }
    }
    return t[0];
}


void zigzag(struct Node* r) {
    if(!r) return;

    struct Node* q[100];
    int f=0,b=0,rev=0;

    q[b++] = r;

    while(f<b) {
        int size = b-f;
        int temp[100];

        for(int i=0;i<size;i++) {
            struct Node* cur = q[f++];
            temp[i] = cur->data;

            if(cur->left) q[b++] = cur->left;
            if(cur->right) q[b++] = cur->right;
        }

        if(rev) {
            for(int i=size-1;i>=0;i--) printf("%d ", temp[i]);
        } else {
            for(int i=0;i<size;i++) printf("%d ", temp[i]);
        }

        rev = !rev;
    }
}

int main() {
    int n; scanf("%d",&n);
    int a[n];
    for(int i=0;i<n;i++) scanf("%d",&a[i]);

    struct Node* root = build(a,n);
    zigzag(root);
}
```
