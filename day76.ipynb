```python
%%writefile day76.c
#include <stdio.h>

int main() {
    int arr[] = {15, -2, 2, -8, 1, 7, 10, 23};
    int n = 8;

    int maxLen = 0;
    int sum = 0;

    int hash[1000] = {0};

    for (int i = 0; i < 1000; i++) {
        hash[i] = -1;
    }

    for (int i = 0; i < n; i++) {
        sum += arr[i];

        if (sum == 0) {
            maxLen = i + 1;
        }

        if (hash[sum + 500] != -1) {
            int len = i - hash[sum + 500];
            if (len > maxLen) {
                maxLen = len;
            }
        } else {
            hash[sum + 500] = i;
        }
    }

    printf("%d", maxLen);

    return 0;
}
```
