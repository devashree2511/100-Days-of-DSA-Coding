```python
%%writefile day55.c
#include <stdio.h>
#include <stdlib.h>

struct Node {
    int data;
    struct Node *left, *right,
};

struct Node* newNode(int d) {
    struct Node* n = (struct Node*)malloc(sizeof(struct Node));
    n->data = d;
    n->left = n->right = NULL;
    return n;
}


struct Node* build(int a[], int n) {
    struct Node* t[n];
    for (int i = 0; i < n; i++)
        t[i] = (a[i] == -1) ? NULL : newNode(a[i]);

    for (int i = 0; i < n; i++) {
        if (t[i]) {
            if (2*i+1 < n) t[i]->left = t[2*i+1];
            if (2*i+2 < n) t[i]->right = t[2*i+2];
        }
    }
    return t[0];
}


void vertical(struct Node* r) {
    int map[100][100], cnt[100]={0};
    int off = 50, min=0, max=0;

    struct { struct Node* n; int hd; } q[100];
    int f=0, b=0;

    q[b++] = (typeof(q[0])){r,0};

    while (f<b) {
        struct Node* cur = q[f].n;
        int hd = q[f++].hd;

        int i = hd + off;
        map[i][cnt[i]++] = cur->data;

        if (hd<min) min=hd;
        if (hd>max) max=hd;

        if (cur->left)  q[b++] = (typeof(q[0])){cur->left, hd-1};
        if (cur->right) q[b++] = (typeof(q[0])){cur->right, hd+1};
    }

    for (int i=min; i<=max; i++) {
        for (int j=0; j<cnt[i+off]; j++)
            printf("%d ", map[i+off][j]);
        printf("\n");
    }
}

int main() {
    int n; scanf("%d",&n);
    int a[n];
    for(int i=0;i<n;i++) scanf("%d",&a[i]);

    struct Node* root = build(a,n);
    vertical(root);
}
```
