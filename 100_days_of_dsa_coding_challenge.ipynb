```python
%%writefile array_insert.c
#include <stdio.h>

int main() {
    int n, pos, x;
    int arr[100];

    scanf("%d", &n);

    for(int i = 0; i < n; i++) {
        scanf("%d", &arr[i]);
    }

    scanf("%d", &pos);
    scanf("%d", &x);

    for(int i = n; i >= pos; i--) {
        arr[i] = arr[i - 1];
    }

    arr[pos - 1] = x;

    for(int i = 0; i <= n; i++) {
        printf("%d ", arr[i]);
    }

    return 0;
}
```

```python
%%writefile day2_array_delete.c
#include <stdio.h>

int main() {
    int n, pos;
    int arr[100];
    int i = 0;

    // Read number of elements
    scanf("%d", &n);

    // Read array elements
    while (i < n) {
        scanf("%d", &arr[i]);
        i++;
    }

    // Read position to delete
    scanf("%d", &pos);

    // Shift elements left
    i = pos - 1;
    while (i < n - 1) {
        arr[i] = arr[i + 1];
        i++;
    }

    // Print updated array
    i = 0;
    while (i < n - 1) {
        printf("%d ", arr[i]);
        i++;
    }

    return 0;
}
```
