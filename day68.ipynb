```python
%%writefile day68.c
#include <stdio.h>
#include <stdlib.h>

int adj[100][100];
int indeg[100];
int queue[100];

int main() {
    int n, m, u, v;
    scanf("%d%d", &n, &m);

    for (int i = 0; i < n; i++) {
        indeg[i] = 0;
        for (int j = 0; j < n; j++)
            adj[i][j] = 0;
    }

    for (int i = 0; i < m; i++) {
        scanf("%d%d", &u, &v);
        adj[u][v] = 1;
        indeg[v]++;
    }

    int front = 0, rear = -1;

    for (int i = 0; i < n; i++)
        if (indeg[i] == 0)
            queue[++rear] = i;

    while (front <= rear) {
        int curr = queue[front++];
        printf("%d ", curr);

        for (int i = 0; i < n; i++) {
            if (adj[curr][i]) {
                indeg[i]--;
                if (indeg[i] == 0)
                    queue[++rear] = i;
            }
        }
    }

    return 0;
}
```
