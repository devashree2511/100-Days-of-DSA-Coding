```python
%%writefile remove.c
#include <stdio.h>

int main() {
    int n;
    scanf("%d", &n);

    int arr[n];
    for (int i = 0; i < n; i++) {
        scanf("%d", &arr[i]);
    }

    // If array is not empty
    if (n > 0) {
        // First element is always unique
        printf("%d ", arr[0]);

        for (int i = 1; i < n; i++) {
            if (arr[i] != arr[i - 1]) {
                printf("%d ", arr[i]);
            }
        }
    }

    return 0;
}
```
