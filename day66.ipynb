```python
%%writefile day66.c
#include <stdio.h>
#include <stdlib.h>

struct Node { int v; struct Node* next; };

struct Node* adj[100];
int vis[100], rec[100];

struct Node* newNode(int v){
    struct Node* n = (struct Node*)malloc(sizeof(struct Node));
    n->v = v; n->next = NULL;
    return n;
}

// Directed edge
void addEdge(int u, int v){
    struct Node* n = newNode(v);
    n->next = adj[u];
    adj[u] = n;
}

int dfs(int v){
    vis[v] = 1;
    rec[v] = 1;

    for(struct Node* t = adj[v]; t; t = t->next){
        if(!vis[t->v] && dfs(t->v)) return 1;
        else if(rec[t->v]) return 1;
    }

    rec[v] = 0;
    return 0;
}

int main(){
    int n, m, u, v;
    scanf("%d%d", &n, &m);

    for(int i=0;i<n;i++)
        adj[i]=NULL, vis[i]=rec[i]=0;

    for(int i=0;i<m;i++){
        scanf("%d%d",&u,&v);
        addEdge(u,v);
    }

    for(int i=0;i<n;i++){
        if(!vis[i] && dfs(i)){
            printf("YES");
            return 0;
        }
    }

    printf("NO");
    return 0;
}
```
