```python
%%writefile day18.c
#include <stdio.h>

int main() {
    int n, k;

    scanf("%d", &n);

    int arr[n], temp[n];
    int i = 0;
    while(i < n) {
        scanf("%d", &arr[i]);
        i++;
    }
    scanf("%d", &k);
    k = k % n;

    i = 0;
    while(i < n) {
        temp[(i + k) % n] = arr[i];
        i++;
    }
    i = 0;
    while(i < n) {
        printf("%d ", temp[i]);
        i++;
    }

    return 0;
}
```
