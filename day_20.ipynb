```python
%%writefile day20.c
#include <stdio.h>
#include <stdlib.h>

int main() {
    int n;
    scanf("%d", &n);

    int arr[n];
    for(int i = 0; i < n; i++) {
        scanf("%d", &arr[i]);
    }

    int count = 0;
    int prefix_sum = 0;

    int hash[10001] = {0};
    hash[0] = 1;

    for(int i = 0; i < n; i++) {
        prefix_sum += arr[i];

        if(hash[prefix_sum] != 0) {
            count += hash[prefix_sum];
        }

        hash[prefix_sum]++;
    }

    printf("%d", count);

    return 0;
}
```
