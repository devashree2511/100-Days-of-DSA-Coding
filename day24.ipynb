```python
%%writefile day24.c
#include <stdio.h>
#include <stdlib.h>

struct node{
    int data;
    struct node* next;
};

int main(){
    int n, key, i = 0;
    scanf("%d", &n);

    struct node *head = NULL, *temp = NULL, *newnode;

    while(i < n){
        newnode = (struct node*)malloc(sizeof(struct node));
        scanf("%d", &newnode->data);
        newnode->next = NULL;

        if(head == NULL)
            head = temp = newnode;
        else{
            temp->next = newnode;
            temp = newnode;
        }
        i++;
    }

    scanf("%d", &key);

    struct node *cur = head, *prev = NULL;

    while(cur != NULL){
        if(cur->data == key){
            if(prev == NULL)
                head = cur->next;
            else
                prev->next = cur->next;
            free(cur);
            break;
        }
        prev = cur;
        cur = cur->next;
    }

    // Print list
    temp = head;
    while(temp != NULL){
        printf("%d ", temp->data);
        temp = temp->next;
    }

    return 0;
}
```
