```python
%%writefile day32.c
#include <stdio.h>
#define STACK_SIZE 1000

int stack[STACK_SIZE];
int top = -1;

void push(int value) {
    if (top == STACK_SIZE - 1) {
        printf("Stack Overflow\n");
        return;
    }
    stack[++top] = value;
}

void pop() {
    if (top == -1) {
        printf("Stack Underflow\n");
        return;
    }
    top--;
}

int main() {
    int n, m, i, value;

    scanf("%d", &n);

    for (i = 0; i < n; i++) {
        scanf("%d", &value);
        push(value);
    }

    scanf("%d", &m);

    for (i = 0; i < m; i++) {
        pop();
    }

    for (i = top; i >= 0; i--) {
        printf("%d ", stack[i]);
    }

    return 0;
}
```
