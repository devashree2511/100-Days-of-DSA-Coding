```python
%%writefile day83.c
#include <stdio.h>

int main() {
    int n, i, j, min_idx, temp;


    scanf("%d", &n);

    int arr[n];


    for(i = 0; i < n; i++) {
        scanf("%d", &arr[i]);
    }


    for(i = 0; i < n - 1; i++) {
        min_idx = i;

        for(j = i + 1; j < n; j++) {
            if(arr[j] < arr[min_idx]) {
                min_idx = j;
            }
        }

        // Swap
        temp = arr[i];
        arr[i] = arr[min_idx];
        arr[min_idx] = temp;
    }

    for(i = 0; i < n; i++) {
        printf("%d ", arr[i]);
    }

    return 0;
}
```
