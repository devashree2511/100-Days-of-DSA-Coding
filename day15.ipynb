```python
%%writefile day15.c
#include <stdio.h>

int main() {
    int m, n;

    scanf("%d %d", &m, &n);

    int matrix[m][n];

    int i = 0, j;

    while(i < m) {
        j = 0;
        while(j < n) {
            scanf("%d", &matrix[i][j]);
            j++;
        }
        i++;
    }

    int sum = 0;
    i = 0;

    while(i < m && i < n) {
        sum = sum + matrix[i][i];
        i++;
    }

    printf("%d", sum);

    return 0;
}
```
