```python
%%writefile day55.c
#include <stdio.h>
#include <stdlib.h>

struct Node {
    int data;
    struct Node *left, *right;
};

struct Node* newNode(int d) {
    struct Node* n = malloc(sizeof(struct Node));
    n->data = d;
    n->left = n->right = NULL;
    return n;
}

struct Node* build(int a[], int n) {
    struct Node* t[n];
    for(int i=0;i<n;i++)
        t[i] = (a[i]==-1)? NULL : newNode(a[i]);

    for(int i=0;i<n;i++) {
        if(t[i]) {
            if(2*i+1<n) t[i]->left = t[2*i+1];
            if(2*i+2<n) t[i]->right = t[2*i+2];
        }
    }
    return t[0];
}

void rightView(struct Node* r) {
    struct Node* q[100];
    int f=0,b=0;

    q[b++] = r;

    while(f<b) {
        int n = b-f;

        for(int i=0;i<n;i++) {
            struct Node* t = q[f++];

            if(i == n-1)
                printf("%d ", t->data);

            if(t->left) q[b++] = t->left;
            if(t->right) q[b++] = t->right;
        }
    }
}

int main() {
    int n; scanf("%d",&n);
    int a[n];
    for(int i=0;i<n;i++) scanf("%d",&a[i]);

    struct Node* root = build(a,n);
    rightView(root);
}
```
