```python
%%writefile day69.c
#include <stdio.h>
#include <limits.h>

#define MAX 100

int n, adj[MAX][MAX];

int minDist(int dist[], int visited[]) {
    int min = INT_MAX, index = -1;
    for (int i = 0; i < n; i++) {
        if (!visited[i] && dist[i] < min) {
            min = dist[i];
            index = i;
        }
    }
    return index;
}

void dijkstra(int src) {
    int dist[MAX], visited[MAX];

    for (int i = 0; i < n; i++) {
        dist[i] = INT_MAX;
        visited[i] = 0;
    }

    dist[src] = 0;

    for (int i = 0; i < n - 1; i++) {
        int u = minDist(dist, visited);
        visited[u] = 1;

        for (int v = 0; v < n; v++) {
            if (!visited[v] && adj[u][v] &&
                dist[u] != INT_MAX &&
                dist[u] + adj[u][v] < dist[v]) {
                dist[v] = dist[u] + adj[u][v];
            }
        }
    }

    for (int i = 0; i < n; i++)
        printf("%d ", dist[i]);
}

int main() {
    int m, u, v, w, src;
    scanf("%d%d", &n, &m);


    for (int i = 0; i < n; i++)
        for (int j = 0; j < n; j++)
            adj[i][j] = 0;

    for (int i = 0; i < m; i++) {
        scanf("%d%d%d", &u, &v, &w);
        adj[u][v] = w;
        adj[v][u] = w;
    }

    scanf("%d", &src);

    dijkstra(src);

    return 0;
}
```
