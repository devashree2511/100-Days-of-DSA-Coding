```python
%%writefile day34.c
#include <stdio.h>
#include <stdlib.h>
#include <ctype.h>

struct Node {
    int data;
    struct Node* next;
};

struct Node* top = NULL;

void push(int val) {
    struct Node* newNode = (struct Node*)malloc(sizeof(struct Node));
    if (!newNode) {
        printf("Heap Overflow\n");
        exit(1);
    }
    newNode->data = val;
    newNode->next = top;
    top = newNode;
}

int pop() {
    if (top == NULL) {
        printf("Stack Underflow\n");
        exit(1);
    }
    struct Node* temp = top;
    int val = temp->data;
    top = top->next;
    free(temp);
    return val;
}

int evaluatePostfix(char* exp) {
    for (int i = 0; exp[i] != '\0'; i++) {
        if (exp[i] == ' ') continue;

        if (isdigit(exp[i])) {
            push(exp[i] - '0');
        }
        else {
            int val1 = pop(); // Right operand
            int val2 = pop(); // Left operand

            switch (exp[i]) {
                case '+': push(val2 + val1); break;
                case '-': push(val2 - val1); break;
                case '*': push(val2 * val1); break;
                case '/': push(val2 / val1); break;
            }
        }
    }
    return pop();
}

int main() {
    char exp[100];
    printf("Enter postfix expression (e.g., 231*+9-): ");
    scanf("%[^\n]s", exp); // Reads string with spaces

    printf("Result: %d\n", evaluatePostfix(exp));
    return 0;
}
```
