```python
%%writefile day25.c
#include <stdio.h>
#include <stdlib.h>

struct node{
    int data;
    struct node* next;
};

int main(){
    int n, i = 0, key, count = 0;
    scanf("%d", &n);

    struct node *head = NULL, *temp = NULL, *newnode;

    while(i < n){
        newnode = (struct node*)malloc(sizeof(struct node));
        scanf("%d", &newnode->data);
        newnode->next = NULL;

        if(head == NULL)
            head = temp = newnode;
        else{
            temp->next = newnode;
            temp = newnode;
        }
        i++;
    }

    scanf("%d", &key);

    temp = head;
    while(temp != NULL){
        if(temp->data == key)
            count++;
        temp = temp->next;
    }

    printf("%d", count);

    return 0;
}
```
