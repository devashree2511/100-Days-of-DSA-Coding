```python
%%writefile symm.c
#include <stdio.h>

int main() {
    int m, n;
    scanf("%d %d", &m, &n);

    int matrix[m][n];
    int i = 0, j;

    while (i < m) {
        j = 0;
        while (j < n) {
            scanf("%d", &matrix[i][j]);
            j++;
        }
        i++;
    }


    if (m != n) {
        printf("Not a Symmetric Matrix");
        return 0;
    }


    i = 0;
    int symmetric = 1;

    while (i < m) {
        j = 0;
        while (j < n) {
            if (matrix[i][j] != matrix[j][i]) {
                symmetric = 0;
                break;
            }
            j++;
        }
        if (symmetric == 0)
            break;
        i++;
    }

    if (symmetric == 1)
        printf("Symmetric Matrix");
    else
        printf("Not a Symmetric Matrix");

    return 0;
}
```
