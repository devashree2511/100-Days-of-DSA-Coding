```python
%%writefile day78.c
#include <stdio.h>
#define INF 9999

int main() {
    int n, m;
    scanf("%d %d", &n, &m);

    int graph[100][100];

    for (int i = 1; i <= n; i++) {
        for (int j = 1; j <= n; j++) {
            graph[i][j] = INF;
        }
    }

    int u, v, w;

    for (int i = 0; i < m; i++) {
        scanf("%d %d %d", &u, &v, &w);
        graph[u][v] = w;
        graph[v][u] = w;
    }

    int visited[100] = {0};
    int totalWeight = 0;

    visited[1] = 1;

    for (int count = 1; count < n; count++) {
        int min = INF;
        int x = -1, y = -1;

        for (int i = 1; i <= n; i++) {
            if (visited[i]) {
                for (int j = 1; j <= n; j++) {
                    if (!visited[j] && graph[i][j] < min) {
                        min = graph[i][j];
                        x = i;
                        y = j;
                    }
                }
            }
        }

        visited[y] = 1;
        totalWeight += min;
    }

    printf("%d", totalWeight);

    return 0;
}
```
