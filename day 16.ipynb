```python
%%writefile day16.c
#include <stdio.h>

int main() {
    int n;

    scanf("%d", &n);

    int arr[n];
    int visited[n];

    int i = 0;

    while(i < n) {
        scanf("%d", &arr[i]);
        visited[i] = 0;
        i++;
    }

    i = 0;

    while(i < n) {

        if(visited[i] == 1) {
            i++;
            continue;
        }

        int count = 1;
        int j = i + 1;

        while(j < n) {
            if(arr[i] == arr[j]) {
                count++;
            }
            j++;
        }

        printf("%d:%d ", arr[i], count);
        i++;
    }

    return 0;
}
```
