```python
%%writefile day2_array_delete.c
#include <stdio.h>

int main() {
    int n, pos;
    int arr[100];
    int i = 0;

    scanf("%d", &n);

    while (i < n) {
        scanf("%d", &arr[i]);
        i++;
    }

    scanf("%d", &pos);

    i = pos - 1;
    while (i < n - 1) {
        arr[i] = arr[i + 1];
        i++;
    }

    i = 0;
    while (i < n - 1) {
        printf("%d ", arr[i]);
        i++;
    }

    return 0;
}
```
